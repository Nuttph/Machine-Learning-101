```python
def square(n):
    for i in range(n):
        yield i * i
```

```python
square(3)
```

```python
for i in square(3):
    print(i)
```

```python
def my_generator():
    yield 1
    yield 2
    yield 3
```

```python
gen = my_generator()
gen
```

```python
try:
    next(gen)
except StopIteration:
    print("Generator is exhausted")
```

```python
# อ่านไฟล์ทีละบรรทัดโดยใช้ generator
# ประหยัดหน่วยความจำเมื่อไฟล์มีขนาดใหญ่
file_path = 'example.txt'
for line in open(file_path):
    print(line.strip())     
```

```python
data = [10, 20, 30, 40, 50]
def list_iterator(data):
    for item in data:
        yield item
```

```python
iter_obj = list_iterator(data)
print(next(iter_obj))
print(next(iter_obj))
print(next(iter_obj))
print(next(iter_obj))
print(next(iter_obj))
# print(next(iter_obj))
```
