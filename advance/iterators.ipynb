```python
interator = iter(my_list)
```

```python
try:
    print(next(interator))
except StopIteration:
    print("End of Iterator reached.")
```

```python
my_string = "Hello"
string_iterator = iter(my_string)
try:
    while True:
        char = next(string_iterator)
        print(char)
except StopIteration:
    print("End of String Iterator reached.")
```

```python
data = [10, 20, 30, 40, 50]
list_iterator = iter(data)
```

```python
try:
    print(next(list_iterator))
except StopIteration:
    print("End of Iterator reached.")
```
