```python
#function copy
def welcome():
    return "Hello world!"
welcome()
```

```python
wel = welcome
wel()

del welcome #deleting original function
```

```python
def repeat(n):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(n):
                func(*args, **kwargs)
        return wrapper
    return decorator
```

```python
# wel()
@repeat(10)
def greet():
    print("Hello world")
```

```python
greet()
```

```python
def lop(func):
    def wrapper():
        print("Before calling the function")
        func()
        print("After calling the function")
    return wrapper
```

```python
@lop
def say_hello():
    print("Hello!")
```

```python
say_hello()
```
