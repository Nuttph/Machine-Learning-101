```python
import pandas as pd
```

```python
# Series
data = [10, 20, 30, 40, 50]
series = pd.Series(data)
print("Pandas Series:\n", series)
```

```python
data = [10,20,30,40,50]
index = ['a','b','c','d','e']
series = pd.Series(data,index=index)
print("\nPandas Series with custom index:\n", series)
```
